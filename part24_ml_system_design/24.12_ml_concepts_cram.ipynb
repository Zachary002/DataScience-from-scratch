**English**: **A few core concepts you must nail in one sentence (remember the conclusion, then verify with code)**:
- **中文**:**偏差-方差权衡**:模型误差 = 偏差² + 方差 + 噪声。**欠拟合=高偏差**(模型太简单,训练测试都差);**过拟合=高方差**(模型太复杂,训练好测试崩)。要在两者间找平衡(正则化、更多数据、合适复杂度)。
  **Bias-variance tradeoff**: model error = bias² + variance + noise. **Underfitting = high bias** (model too simple, both train and test poor); **overfitting = high variance** (model too complex, train good test collapses). Balance between them (regularization, more data, right complexity).
- **中文**:**L1 vs L2 正则化**:都惩罚大权重防过拟合。**L1(Lasso)让权重变稀疏(很多精确为 0)→ 自带特征选择**;**L2(Ridge)让权重变小但不为 0**。要选特征用 L1,只想缩小用 L2。
  **L1 vs L2 regularization**: both penalize large weights to prevent overfitting. **L1 (Lasso) makes weights sparse (many exactly 0) → built-in feature selection**; **L2 (Ridge) makes weights small but nonzero**. Use L1 for feature selection, L2 for shrinkage.
- **中文**:**Bagging vs Boosting**:都是集成。**Bagging(如随机森林)并行训练多个模型再平均→降方差**(治过拟合);**Boosting(如 XGBoost)串行, 每个新模型纠正前面的错误→降偏差**(治欠拟合, 但更易过拟合)。
  **Bagging vs Boosting**: both ensembles. **Bagging (e.g. random forest) trains many models in parallel then averages → reduces variance** (fixes overfitting); **Boosting (e.g. XGBoost) is sequential, each new model corrects prior errors → reduces bias** (fixes underfitting, but more prone to overfit).

> 💡 **高频 ML 概念 Q&A 速查 / Frequent ML concept Q&A cheat-sheet（★★★ 逢面必考)**
> **中文**:
> - **偏差-方差**? 误差=偏差²+方差+噪声; 欠拟合高偏差(都差)、过拟合高方差(训练好测试崩); 正则化/更多数据/调复杂度平衡。
> - **过拟合怎么办**? 更多数据、正则化(L1/L2/dropout)、简化模型、交叉验证、早停、集成、数据增强。
> - **L1 vs L2**? L1 稀疏(特征选择), L2 缩小不置零; L1 对应拉普拉斯先验、L2 高斯先验。
> - **为什么交叉验证**? 单次划分方差大/浪费数据; k 折让每个样本都当过验证, 估计更稳、更充分利用数据; 时序用 TimeSeriesSplit。
> - **Bagging vs Boosting**? bagging 并行降方差(随机森林), boosting 串行降偏差(XGB/LightGBM); RF 抗过拟合稳、boosting 更准但需调参。
> - **准确率的问题**? 不平衡下无用(全预测多数类); 用 precision/recall/F1/PR-AUC/ROC-AUC(接 24.5)。
> - **precision vs recall**? precision=预测为正里真正的比例(查准), recall=真正例里被找到的比例(查全); 权衡靠阈值; F1 是调和平均。
> - **ROC-AUC vs PR-AUC**? AUC=排序能力(随机正例分>随机负例的概率); 极不平衡时 PR-AUC 更敏感; AUC 不衡量校准(接 24.4)。
> - **生成式 vs 判别式**? 判别式建 P(y|x)(直接分类, LR/SVM/树), 生成式建 P(x,y)/P(x|y)(朴素贝叶斯/GMM), 判别式通常分类更准。
> - **参数 vs 非参数**? 参数=固定参数量(LR/线性), 非参数=随数据增长(KNN/决策树/核方法), 后者更灵活但更易过拟合。
> - **梯度下降类型**? 批量(全数据, 稳但慢)、随机SGD(单样本, 快但抖)、mini-batch(折中, 标配); 加动量/Adam 加速。
> - **梯度消失/爆炸**? 深网反向传播梯度连乘→消失(sigmoid/tanh 饱和)或爆炸; 解: ReLU、残差连接、BatchNorm、梯度裁剪、合适初始化。
> - **判断特征重要性**? 树的 importance/SHAP(接 Part 8), 但相关≠因果(接 19); permutation importance 更可靠。
> - **维度灾难**? 高维下距离趋同、数据稀疏、需指数级样本; 解: 降维(PCA)、特征选择、正则化。
> - **数据泄漏**? 用了预测时拿不到的信息(预处理在划分前/未来信息/标签代理)→评估虚高上线崩(接 22.2); 用 Pipeline 防。
> - **类别不平衡**? 重采样(SMOTE)、类别权重、阈值调整、异常检测、用对指标(接 24.5)。
> - **归一化 vs 标准化**? 归一化到[0,1](MinMax), 标准化到均值0方差1(Standard); 距离/梯度类模型需要, 树不需要。
> - **判别模型欠拟合还是过拟合**? 看训练vs验证: 都差=欠拟合(加复杂度/特征), 训练好验证差=过拟合(正则/更多数据)。
> **English**:
> - **Bias-variance**? Error = bias² + variance + noise; underfit high bias (both poor), overfit high variance (train good test collapses); balance via regularization/more data/complexity.
> - **Fix overfitting**? More data, regularization (L1/L2/dropout), simpler model, cross-validation, early stopping, ensembles, data augmentation.
> - **L1 vs L2**? L1 sparse (feature selection), L2 shrinks not zeros; L1 ~ Laplace prior, L2 ~ Gaussian prior.
> - **Why cross-validation**? A single split is high-variance/wasteful; k-fold uses every sample as validation, more stable estimate, fuller use of data; time series uses TimeSeriesSplit.
> - **Bagging vs Boosting**? Bagging parallel reduces variance (random forest), boosting sequential reduces bias (XGB/LightGBM); RF is robust, boosting more accurate but needs tuning.
> - **Problem with accuracy**? Useless under imbalance (predict majority); use precision/recall/F1/PR-AUC/ROC-AUC (per 24.5).
> - **Precision vs recall**? Precision = of predicted positives how many are true (exactness), recall = of true positives how many are found (completeness); trade off via threshold; F1 is the harmonic mean.
> - **ROC-AUC vs PR-AUC**? AUC = ranking ability (probability a random positive scores > a random negative); PR-AUC more sensitive under extreme imbalance; AUC doesn't measure calibration (per 24.4).
> - **Generative vs discriminative**? Discriminative models P(y|x) (direct classification, LR/SVM/trees), generative models P(x,y)/P(x|y) (naive Bayes/GMM); discriminative usually classifies better.
> - **Parametric vs non-parametric**? Parametric = fixed number of parameters (LR/linear), non-parametric = grows with data (KNN/decision trees/kernels), the latter more flexible but more overfit-prone.
> - **Gradient descent types**? Batch (all data, stable but slow), SGD (one sample, fast but noisy), mini-batch (middle ground, standard); add momentum/Adam to speed up.
> - **Vanishing/exploding gradients**? Deep-net backprop multiplies gradients → vanish (sigmoid/tanh saturation) or explode; fixes: ReLU, residual connections, BatchNorm, gradient clipping, proper initialization.
> - **Feature importance**? Tree importance/SHAP (per Part 8), but correlation ≠ causation (per 19); permutation importance more reliable.
> - **Curse of dimensionality**? In high dimensions distances converge, data is sparse, needs exponential samples; fixes: dimensionality reduction (PCA), feature selection, regularization.
> - **Data leakage**? Using info unavailable at prediction (preprocessing before split / future info / label proxy) → inflated evaluation, live collapse (per 22.2); prevent with Pipelines.
> - **Class imbalance**? Resampling (SMOTE), class weights, threshold tuning, anomaly detection, use the right metrics (per 24.5).
> - **Normalization vs standardization**? Normalize to [0,1] (MinMax), standardize to mean 0 var 1 (Standard); distance/gradient models need it, trees don't.
> - **Underfit or overfit**? Check train vs validation: both poor = underfit (add complexity/features), train good validation poor = overfit (regularize/more data).


```python

# ============================================================
# 演示①:偏差-方差权衡(多项式次数)/ demo ①: bias-variance tradeoff (polynomial degree)
# 中文:同一份带噪声的数据, 拟合不同复杂度的模型。太简单=欠拟合(训练测试都差), 太复杂=过拟合(训练好测试崩)。
# English: same noisy data, fit models of different complexity. Too simple = underfit (both poor), too complex = overfit (train good test collapses).
# ============================================================
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error
np.random.seed(0)
X=np.sort(np.random.rand(30,1),axis=0); y=np.sin(2*np.pi*X[:,0])+0.15*np.random.randn(30)  # 真实是正弦+噪声
Xtr,ytr,Xte,yte=X[:20],y[:20],X[20:],y[20:]
print("偏差-方差权衡(多项式次数越高越复杂)/ bias-variance (higher degree = more complex):")
for d,label in [(1,"欠拟合 underfit (高偏差)"),(4,"较好 good"),(15,"过拟合 overfit (高方差)")]:
    m=make_pipeline(PolynomialFeatures(d),LinearRegression()).fit(Xtr,ytr)
    tr=mean_squared_error(ytr,m.predict(Xtr)); te=mean_squared_error(yte,m.predict(Xte))
    print(f"  degree={d:>2} {label:28}: 训练MSE {tr:6.3f}  测试MSE {te:10.3f}")
print("→ degree 1: 训练测试都差(高偏差); degree 15: 训练极好但测试爆炸(高方差过拟合)。要找中间的平衡点。")
```

```python

# ============================================================
# 演示②:L1(Lasso, 稀疏)vs L2(Ridge, 缩小)正则化 / demo ②: L1 (sparse) vs L2 (shrink) regularization
# 中文:真实权重是稀疏的(只有3个非零)。L1 能把无关特征的权重精确压到0(特征选择), L2 只缩小不置零。
# English: true weights are sparse (only 3 nonzero). L1 drives irrelevant weights exactly to 0 (feature selection); L2 shrinks but doesn't zero.
# ============================================================
from sklearn.linear_model import Lasso, Ridge
np.random.seed(1)
X=np.random.randn(200,10); w_true=np.array([3,-2,0,0,0,1.5,0,0,0,0]); y=X@w_true+0.5*np.random.randn(200)
l1=Lasso(alpha=0.1).fit(X,y); l2=Ridge(alpha=10).fit(X,y)
fmt=lambda coef: [round(float(c),2) for c in coef]
print("真实权重  true w :", w_true.tolist())
print("L1 (Lasso) 权重  :", fmt(l1.coef_), f"→ {int(np.sum(np.abs(l1.coef_)<1e-4))} 个精确为0 (稀疏=自带特征选择)")
print("L2 (Ridge) 权重  :", fmt(l2.coef_), f"→ {int(np.sum(np.abs(l2.coef_)<1e-4))} 个为0 (只缩小, 不置零)")
print("→ L1 精确复现了稀疏结构(把无关的5个特征压到0), L2 让所有权重都变小但没有一个是0。要特征选择用 L1。")
```

```python

# ============================================================
# 演示③:Bagging(降方差)vs Boosting(降偏差)+ 可视化 / demo ③: bagging vs boosting + viz
# 中文:同一份数据。单棵深树容易过拟合(高方差); bagging(随机森林)平均多棵树降方差; boosting 串行纠错降偏差。
# English: same data. A single deep tree overfits (high variance); bagging (random forest) averages trees to reduce variance; boosting sequentially corrects errors to reduce bias.
# ============================================================
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.datasets import make_classification
from sklearn.model_selection import cross_val_score
Xc,yc=make_classification(n_samples=1500,n_features=20,n_informative=8,random_state=0)
models={"单棵深树\n(高方差)":DecisionTreeClassifier(max_depth=None,random_state=0),
        "Bagging\n随机森林\n(降方差)":RandomForestClassifier(n_estimators=100,random_state=0),
        "Boosting\nGBDT\n(降偏差)":GradientBoostingClassifier(n_estimators=100,random_state=0)}
scores={k:cross_val_score(m,Xc,yc,cv=5).mean() for k,m in models.items()}
fig,ax=plt.subplots(1,2,figsize=(14,5))
ax[0].bar(list(scores.keys()),list(scores.values()),color=["#C44E52","#55A868","#4C72B0"])
for i,v in enumerate(scores.values()): ax[0].text(i,v+0.005,f"{v:.3f}",ha="center",fontsize=11,weight="bold")
ax[0].set_ylim(0.8,0.95); ax[0].set_ylabel("5折交叉验证准确率"); ax[0].set_title("集成: bagging 降方差, boosting 降偏差")
# bias-variance 概念图 / bias-variance concept
ax[1].axis("off"); ax[1].set_title("偏差-方差权衡(总误差的 U 形)",fontsize=12,weight="bold")
comp=np.linspace(0,10,100); bias=(10-comp)**2/12; var=comp**2/12; total=bias+var+1
ax[1]=fig.add_subplot(122); ax[1].plot(comp,bias,label="偏差² (欠拟合)",color="#4C72B0")
ax[1].plot(comp,var,label="方差 (过拟合)",color="#C44E52"); ax[1].plot(comp,total,label="总误差",color="k",lw=2)
ax[1].axvline(comp[total.argmin()],ls="--",color="#55A868",label="最优复杂度")
ax[1].set_xlabel("模型复杂度 →"); ax[1].set_ylabel("误差"); ax[1].set_title("偏差-方差: 总误差是 U 形"); ax[1].legend(fontsize=8); ax[1].set_yticks([])
plt.tight_layout(); plt.savefig("/tmp/sd12_viz.png",dpi=80); plt.show()
print("左:随机森林(bagging)和 GBDT(boosting)都比单棵树好; 右:总误差=偏差²+方差, 随复杂度增加呈 U 形, 要找最低点")
```
